WAP to calculate area of rectangle, square, circle and triangle. Take
necessary inputs from user.

```python
Shape = input("Enter your shape:")
if Shape == "Square":
    a = int(input("Enter the length of side:"))
    print("Area of Square is:",a*a)
elif Shape == "Rectangle":
    l = int(input("Enter the length:"))
    b = int(input("Enter the breadth:"))
    print("Area of Rectangle is:",l*b)
elif Shape == "Circle":
    r = int(input("Enter the radius:"))
    print("Area of Circle is:",3.14*r*r)
elif Shape == "Triangle":
    b = int(input("Enter the base:"))
    h = int(input("Enter the height:"))
    print("Area of Triangle is:",0.5*b*h)
else:
    print("Invalid Shape or Incorrect Case Used. Please enter the shape name in correct case.")
```

Accept as input the basic salary of a person. His dearness allowance (DA)
is 40% of the basic salary and the house rent allowance (HRA) is 20% of
the basic salary. Calculate and show the Gross salary.

```python
basic_salary = float(input("Enter your salary:"))
dearness_allowance = 0.4 * basic_salary
house_rent_allowance = 0.2 * basic_salary
print("Gross Salary is:",basic_salary + dearness_allowance + house_rent_allowance)
```

Accept as Input the marks obtained by a student in 5 subjects. Show the
Aggregate marks and Percentage marks.

```python
print("Enter the marks of five subjects:")
subject1 = float(input("Enter the marks of first subject:"))
subject2 = float(input("Enter the marks of second subject:"))
subject3 = float(input("Enter the marks of third subject:"))
subject4 = float(input("Enter the marks of fourth subject:"))
subject5 = float(input("Enter the marks of fifth subject:"))
total_marks = subject1 + subject2 + subject3 + subject4 + subject5
percentage = (total_marks / 500) * 100
print("Total Marks Obtained:", total_marks)
print("Percentage:", percentage, "%")
```
