```python

# 1. A college has two lists containing the roll numbers of students enrolled
# in Python and Java courses. Write a Python program to find: (a) students
# enrolled in both courses, (b) students enrolled only in Python, and (c)
# students enrolled in either course.(hint: set(), intersection, difference,
# union)
python={10,12,43,21,45,44,23,42}
java={21,11,22,32,33,42,11}
print("Enrolled in python :",python)
print("Enrolled in java :",java)
both=python.intersection(java)
print("Enrolled in both :",both)
any=python.union(java)
print("Enrolled in any :",any)


```

```python
# 2. Two users have lists of their followers. Write a program to find the
# common followers and the followers who are unique to each user. Also
# determine the total number of unique followers across both users.(hint:
# Union, intersection, difference, len())
s1 = "KISHORE,ARIJIT,SONU,RAVIKISAN"
S2 = "ASHA,LATA,ABHIJIT,SONU"
set1= set(s1.split(","))
set2 = set(S2.split(","))
print(set1)
print(set2)
both_followers = set1.intersection(set2)
print("Common followers:", both_followers)
unique_to_user1 = set1.difference(set2)
print("Followers unique to user 1:", unique_to_user1)
unique_to_user2 = set2.difference(set1)
print("Followers unique to user 2:", unique_to_user2)
total_unique_followers = len(set1.union(set2))
print("Total number of unique followers across both users:", total_unique_followers)
```

```python
# 3. A library maintains a set of book IDs currently issued and another set of
# book IDs that are overdue. Write a program to identify books that are
# both issued and overdue, books that are issued but not overdue, and all
# books involved in either category.(hint: Intersection, difference, union)
Books_Issued={"BK101","BK103","BK105","BK107","BK109","BK105","BK107"}
Books_Overdue={"BK102","BK103","BK105","BK103"}
print("Currently Issued Books : ",Books_Issued)
print("Currently Overdued Books : ",Books_Overdue)
Overdued_Issued_Books=Books_Issued.intersection(Books_Overdue)
print("Overdued Issued Books :",Overdued_Issued_Books)
Not_Overdued_Issued_Books=Books_Issued.difference(Books_Overdue)
print("Non Overdued Issued Book(s) :",Not_Overdued_Issued_Books)
All_Books=Books_Issued.union(Books_Overdue)
print("All Book(s) :",All_Books)

```

```python
# 4. A hospital has sets containing patient IDs who visited the Cardiology,
# Neurology, and Orthopedics departments. Write a program to find
# patients who visited all three departments and patients who visited
# Cardiology but never visited Neurology. (hint: Multiple set intersections,
# difference)
Cardiology_Patients={"PT101","PT102","PT104","PT107"}
Neurology_Patients={"PT102","PT103","PT104","PT105"}
Orthopedics_Patients={"PT101","PT104","PT106","PT107"}
print("Cardiology Patients : ",Cardiology_Patients)
print("Neurology Patients : ",Neurology_Patients)
print("Orthopedics Patients : ",Orthopedics_Patients)
All_Three_Departments=Cardiology_Patients.intersection(Neurology_Patients,Orthopedics_Patients)
print("Patients visited all three departments :",All_Three_Departments)
Cardiology_Only=Cardiology_Patients.difference(Neurology_Patients)
print("Patients who visited Cardiology but never visited Neurology :",Cardiology_Only)
```

```python
# 5. A software project requires the skills Python, Java, SQL, Cloud, and Git.
# Given the skill sets of several employees, write a program to identify
# employees who can independently satisfy all project requirements. If no
# employee qualifies, display the missing skills for each employee. (hint:
# issuperset(), subset, iteration, set difference).
Project_Skills={"Python","Java","SQL","Cloud","Git"}
Employee1_Skills={"Python","Java","SQL"}
Employee2_Skills={"Python","Java","SQL","Cloud","Git"}
Employee3_Skills={"Python","Java","SQL","Cloud"}
Employee4_Skills={"Python","Java","SQL","Git"}
print("Project Skills Required : ",Project_Skills)
print("Employee 1 Skills : ",Employee1_Skills)
print("Employee 2 Skills : ",Employee2_Skills)
print("Employee 3 Skills : ",Employee3_Skills)
print("Employee 4 Skills : ",Employee4_Skills)
Employee_Skills_List=[Employee1_Skills,Employee2_Skills,Employee3_Skills,Employee4_Skills]
for i, skills in enumerate(Employee_Skills_List, start=1):
    if skills.issuperset(Project_Skills):
        print(f"Employee {i} can independently satisfy all project requirements.")
    else:
        missing_skills = Project_Skills.difference(skills)
        print(f"Employee {i} is missing the following skills: {missing_skills}")
```

```python

# 6. A company has a list of employee email addresses. Using set
# comprehension, extract the domain names from all email addresses and
# create a set containing the unique domains. For example,
# user@gmail.com should produce gmail.com. (hint: String operations,
# split(), transformation)
email_addresses = ["user1@gmail.com", "user2@yahoo.com", "user3@gmail.com", "user4@hotmail.com"]
unique_domains = {email.split('@')[1] for email in email_addresses}
print("Unique Domains : ",unique_domains)
```

```python
# 7. A customer database contains mobile numbers as strings, some of which
# have fewer than 10 digits. Using set comprehension, create a set
# containing the last four digits of all valid 10-digit mobile numbers. (hint:
# String slicing, conditions, set comprehension)
Customer_Mobile_Numbers = ["1234567890", "987654321", "5555555555", "12345", "1112223333"]
valid_last_four_digits = {number[-4:] for number in Customer_Mobile_Numbers  if len(number) == 10}
print("Last four digits of valid 10-digit mobile numbers:", valid_last_four_digits)
```

```python

# 8. A list contains the names of employees who know different
# programming languages. Given a list of employee records containing
# their names and skills, use set comprehension to create a set of
# employees who know Python. (hint: Set comprehension with conditions,
# nested data)
employee_records = [["John", ["Python", "Java"]], ["Alice", ["C++", "Python"]], ["Bob", ["JavaScript"]], ["Eve", ["Python", "C#"]]]
python_developers = {name for name, skills in employee_records if "Python" in skills}
print("Python Developers : ",python_developers)
```

```python
# 9. Write a Python program that simulates rolling a pair of dice. Generate
# two random numbers between 1 and 6, display them, and calculate
# their total. Also display whether the player wins when the total is 7 or
# 11.(hint: random.randint(), variables, conditional statements)
import random


dice1 = random.randint(1, 6)
dice2 = random.randint(1, 6)
total = dice1 + dice2

print(f"Die 1: {dice1}")
print(f"Die 2: {dice2}")
print(f"Total: {total}")

if total == 7 or total == 11:
    print("Player wins!")
else:
    print("Player loses.")
```

```python
# 11. A class contains the names of 10 students. Write a program that
# randomly divides them into groups of 5 students each. Every student
# should belong to exactly one group. (hint: random.shuffle(), lists, slicing,
# loops)
class_students = ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah", "Ian", "Jack"]
random.shuffle(class_students)
group1 = class_students[:5]
group2 = class_students[5:10]
print("Group 1:", group1)
print("Group 2:", group2)
```

```python

# 12. Write a Python program that randomly selects a number between 1
# and 100. The user gets 7 attempts to guess the number. After each
# attempt, display whether the guess is too high or too low. (hint:
# random.randint(), loops, conditions)


number = random.randint(1, 100)

for attempts in range(7, 0, -1):
    print(f"You have {attempts} attempts remaining.")
    guess = int(input("Enter your guess: "))
    if guess == number:
        print(f"Congratulations! You guessed the number {number}.")
        break
    elif guess < number:
        print(f"Your guess : {guess} is too low.")
    else:
        print(f"Your guess : {guess} is too high.")
    print("----------------------------------------")
else:
    print(f"Sorry, you ran out of attempts. The number was {number}.")
```

```python

# 13. A weather application needs test data. Write a Python program to
# generate a random temperature between 20.0°C and 40.0°C. Display the
# temperature rounded to 2 decimal places and indicate whether it is
# Cool, Moderate, or Hot. (hint: random.uniform(), round(number,
# ndigits), if-elif-else)
temperature = random.uniform(20.0, 40.0)
temperature = round(temperature, 2)

if temperature < 25.0:
    print(f"Temperature: {temperature}°C - Cool")
elif temperature < 35.0:
    print(f"Temperature: {temperature}°C - Moderate")
else:
    print(f"Temperature: {temperature}°C - Hot")
```

```python
# 14. A company wants to simulate annual salary increments. Generate a
# random increment percentage between 5% and 15% for an employee.
# Given the employee's current salary, calculate the new salary. (hint:
# random.uniform(), arithmetic operations)
current_salary = float(input("Enter the current salary: "))
increment_percentage = random.uniform(5, 15)
new_salary = current_salary * (1 + increment_percentage / 100)
print(f"Current Salary: {current_salary}")
print(f"Increment Percentage: {increment_percentage:.2f}%")
print(f"New Salary: {new_salary:.2f}")
```
